#### **Using _ _ eq _ _ Method to define the behaviour of `==`**

```python
class rectangles:
    ''' Base code for defining the Rectangle Class '''
    def __init__(self,length,breadth):
        self.length = length
        self.breadth = breadth

    def get_area(self):
        return self.length * self.breadth

    def get_perimeter(self):
        return 2*(self.length+self.breadth)

    def __eq__(self,other):
        if self.length * self.breadth == other.length * other.breadth:
            return True
        else:
            return False


r2 = rectangles(100,80)
r3 = rectangles(400,20)
```

```python
r2 == r3
```

```python
print(id(r2))
print(id(r3))
```

#### **Using _ _ gt _ _ Method to define the behaviour of `>`**

```python
class rectangles:
    ''' Base code for defining the Rectangle Class '''
    def __init__(self,length,breadth):
        self.length = length
        self.breadth = breadth

    def get_area(self):
        return self.length * self.breadth

    def get_perimeter(self):
        return 2*(self.length+self.breadth)

    def __eq__(self,other):
        if self.get_area() == other.get_area():
            return True
        else:
            return False

    def __gt__(self,other):
        if self.get_area() > other.get_area():
            return True
        else:
            return False

r2 = rectangles(100,80)
r3 = rectangles(400,30)
```

#### **Using _ _ sub _ _ Method to define the behaviour of `-`**

```python
class rectangles:
    ''' Base code for defining the Rectangle Class '''
    def __init__(self,length,breadth):
        self.length = length
        self.breadth = breadth

    def get_area(self):
        return self.length * self.breadth

    def get_perimeter(self):
        return 2*(self.length+self.breadth)

    def __eq__(self,other):
        if self.get_area() == other.get_area():
            return True
        else:
            return False

    def __gt__(self,other):
        if self.get_area() > other.get_area():
            return True
        else:
            return False

    def __sub__(self,other):
        if self.get_area() > other.get_area():
            return self.get_area() - other.get_area()
        else:
            return other.get_area() - self.get_area()

r2 = rectangles(100,80)
r3 = rectangles(400,30)
```

#### **Using _ _ str _ _ Method to define the behaviour of `print()` function on a Class object**

```python
class rectangles:
    ''' Base code for defining the Rectangle Class '''
    def __init__(self,length,breadth):
        self.length = length
        self.breadth = breadth

    def get_area(self):
        return self.length * self.breadth

    def get_perimeter(self):
        return 2*(self.length+self.breadth)

    def __eq__(self,other):
        if self.get_area() == other.get_area():
            return True
        else:
            return False

    def __gt__(self,other):
        if self.get_area() > other.get_area():
            return True
        else:
            return False

    def __sub__(self,other):
        if self.get_area() > other.get_area():
            return self.get_area() - other.get_area()
        else:
            return other.get_area() - self.get_area()

    def __str__(self):
        return "Rectangle with length "+str(self.length)+" and breadth "+str(self.breadth)

r2 = rectangles(100,80)
r3 = rectangles(400,30)
```

```python
print(r3)
```

```python
print(r2)
```

```python
#----Again tweaking the blueprint to define the two attributes----
class squares(rectangles):
    def __init__(self,side):
        self.length = side
        self.breadth = side

    def __str__(self):
        return "Square with side "+str(self.length)
```

### **Using `super()` function in __init__ method**

```python
class squares(rectangles):
    def __init__(self,side):
        #-------Super() calls the constructor of the Parent Class-----
        super().__init__(side,side)

    def __str__(self):
        return "Square with side "+str(self.length)
```

```python
s1 = squares(90)
s2 = squares(70)
```

```python
print(s2)
```

```python
s2.get_area()
```

```python
s1.get_perimeter()
```
